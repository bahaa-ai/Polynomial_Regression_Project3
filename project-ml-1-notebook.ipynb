```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
m = 100  
X_raw = 2 * np.random.rand(m, 1) - 1
y = 4 * (X_raw ** 2) + 3 * X_raw + 1 + 0.1 * np.random.randn(m, 1)
y = y.squeeze() 
X = np.c_[np.ones(m), X_raw, X_raw**2]

m, n = X.shape

theta = np.zeros(n)
learning_rate = 0.01
epochs = 500
costs = []

 
for epoch in range(epochs):
    predictions = np.dot(X, theta)
    errors = predictions - y
    cost = (1 / (2 * m)) * np.sum(errors ** 2)
    costs.append(cost)
    
    gradient = (1 / m) * np.dot(X.T, errors)
    theta -= learning_rate * gradient

print("GD Final Cost:", costs[-1])
print("GD Final Theta:", theta)

plt.plot(costs)
plt.title("Gradient Descent (GD) - Polynomial Regression")
plt.xlabel("Epochs")
plt.ylabel("Cost")
plt.savefig("gd_cost.png")
plt.show()
```
